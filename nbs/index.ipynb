# Learning Dynamics Under the Lens of Koopman Operator Theory

> This project studies neural networks under the lens of the Koopman Operator Theory (KOT)

#### Research Narrative

Typically KOT is discussed with time-series data, with the expectation that one can learn and linearize the underlying dynamics of a complex system by "watching" the states evolve. 

We are interested in studying each layer of a neural network at stationarity (post-training). Our goal is to characterize each layer's effect on the input. For now, these experiments are limited to Linear+ReLU layers with the same input and output dimensionality. 

#### Approach

To achieve the "time-series" effect, we train an NN on a non-trivial binary classification problem and store the final weights. Then, we feed in-distribution data into the trained network and store the activations of the layer *right* before the ones we are interested in. 

We then "loop" these inputs into the Linear+ReLu layer for $N$ iterations and store all activations. Each neuron is a "state" and each loop is a "timestep".
